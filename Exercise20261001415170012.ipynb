2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙

Python 的 float 使用 IEEE 754 雙精度浮點數格式儲存數值。

因為電腦只能使用有限的位元，所以像 0.1 這種二進位無限循環的小數無法完整儲存，只能保留有限位數，並將後面的位數進行捨入。

IEEE 754 常使用「四捨六入五成雙」（round to nearest, ties to even）的方式進行捨入。

因此 Python 儲存的 0.1 並不是數學上完全精確的 0.1，而是一個非常接近 0.1 的二進位近似值。

### 4. 為什麼 0.1 + 0.2 不等於精確的 0.3？

因為 0.1 和 0.2 在二進位中都是無限循環小數，
電腦只能用有限的位元儲存近似值。

因此兩個近似值相加後，會產生微小的浮點誤差：

0.1 + 0.2 = 0.30000000000000004

若只是一般比較，可以使用 round()：

round(0.1 + 0.2, 1) == 0.3

結果會是 True。

若要判斷兩個浮點數是否「足夠接近」，
可以使用 math.isclose()。

### 5. 在科學或金融計算中是否適合？

不一定，要看用途。

在科學計算中，浮點數本來就可能有微小誤差，
因此使用 `math.isclose()` 設定合理的誤差範圍，
通常比直接使用 `==` 比較浮點數更適合。

`round()` 可以用來控制顯示的小數位數，
但不代表底層的浮點誤差真正消失。

在金融計算中，金額通常要求精確，
不適合單純使用 `float`，
因為誤差累積可能造成金額不正確。

金融計算較適合使用 Python 的 `Decimal`，
因為它可以用十進位方式精確表示金額。
