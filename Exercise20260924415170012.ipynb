## 第三部分：CPython 如何實作這些型別？

| 數學概念 | Python 型別 | CPython 實作 | 主要特點 |
|----------|------------|-------------|----------|
| 自然數 | `int` | PyLongObject | 任意精度整數，可表示很大的整數 |

## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)

```

回答：

1. 為什麼 Python 能表示如此大的整數？
因為 Python 的 int 採用任意精度整數，可以依照數字大小動態增加儲存空間。
2. Python 的 `int` 是否具有固定的位元數限制？
沒有。Python 的 int 不像一般的 32-bit 或 64-bit 整數有固定大小。
3. 什麼是 Arbitrary Precision（任意精度）？
指數字需要多少儲存空間，就可以動態使用多少記憶體，因此可以表示非常大的整數。
4. CPython 使用什麼物件來實作 `int`？
PyLongObject
5. Python 整數大小真正的限制是什麼？
主要受到電腦可用記憶體的限制。
---

再執行：

```python
0.1 + 0.2
```

```python
0.1 + 0.2 == 0.3
```

```python
format(0.1, ".20f")
```

回答：

1. 為什麼結果與數學上的答案不同？
因為電腦使用二進位浮點數表示小數，而 0.1、0.2 無法用有限的二進位位數精確表示，因此會產生微小誤差。
2. 浮點數是否能精確表示所有小數？
不能。只有部分數值可以被精確表示，其他數值只能近似表示。
3. 什麼是浮點數的精度（precision）？
精度是浮點數能夠可靠表示數值的有效位數。Python 的 float 通常約有 15～17 位十進位有效數字。
4. 什麼是浮點數的範圍（range）？
範圍是浮點數可以表示的最小到最大數值大小。Python float 通常可以表示到大約 10^308 的數量級。
5. 精度與範圍有何不同？
精度是「數值能表示得多準確」；範圍是「數值能表示得多大或多小」。

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？
通常使用浮點數，常見有 FP32、FP16、BF16。
2. 什麼是 FP64？
FP64 是 64 位元浮點數，又稱雙精度浮點數，精度高但占用記憶體較多、計算成本較高。
3. 什麼是 FP32？
FP32 是 32 位元浮點數，又稱單精度浮點數，是常見的數值運算與 AI 訓練格式。
4. 什麼是 FP16？
FP16 是 16 位元浮點數，占用空間較小、運算速度快，常用於 AI 訓練與推論。
5. 什麼是 BF16？
BF16 是 16 位元浮點格式（BFloat16）。它的數值範圍接近 FP32，但精度較低，常用於 AI 訓練。
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？
因為模型權重通常包含小數，例如 0.25、-0.013，而 int 只能表示整數；此外浮點格式也更適合神經網路的大量矩陣運算。

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？
表示 Python 的 int 沒有固定的位元數限制，數值變大時可以使用更多記憶體來儲存。
2. 它是否代表可以儲存無限大的整數？
不是。理論上沒有固定的整數上限，但實際上仍受到電腦可用記憶體的限制。
3. CPython 如何實現這項特性？
CPython 使用 PyLongObject 實作 int，將大整數分成多個 digit 儲存。當整數變大時，就配置更多記憶體來增加 digit 的數量。

---

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？
因為 Python 的 int 採用任意精度表示，數字變大時可以動態增加儲存空間，因此沒有固定的 32-bit 或 64-bit 上限。
2. 為什麼 `float` 卻存在範圍限制？
因為 Python 的 float 通常使用固定大小的 64-bit 浮點數表示，所以可表示的數值範圍與精度都是有限的。
3. 為什麼 `float` 可能產生誤差？
因為許多十進位小數無法用有限的二進位位數精確表示，所以電腦只能儲存近似值，因此可能產生浮點誤差。

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。
Python 的整數採用 Arbitrary Precision（任意精度），重視整數運算的正確性；AI 常用的 FP16、BF16 則採用 Limited Precision（有限精度），以較低精度換取更高的運算效率與較少的記憶體使用。
2. 各自的優點與缺點。
Python int 的優點是可以精確表示很大的整數，缺點是數字越大就需要越多記憶體，運算也可能變慢。FP16、BF16 的優點是占用記憶體少、運算速度快，適合大量矩陣運算；缺點是精度較低，可能產生誤差。
3. 為什麼大型語言模型通常選擇較低精度的表示方式？
因為大型模型有數十億甚至更多參數，使用 FP16 或 BF16 可以大幅減少記憶體需求、提高 GPU 運算速度與資料傳輸效率，而且通常仍能維持足夠的模型準確度。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？
會需要更多記憶體、運算速度變慢、耗電量增加、訓練成本提高，也會降低 GPU 可以同時處理的資料量，而對多數 AI 模型來說通常不會帶來相同比例的效果提升。
